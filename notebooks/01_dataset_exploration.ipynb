# 01 - Underwater Dataset Exploration

This notebook explores underwater image datasets (e.g., UIEB, EUVP, UFO-120), analyzing optical underwater phenomena such as:
- **Wavelength-dependent light attenuation** (rapid absorption of red light)
- **Color casting** (dominant green and blue hues)
- **Low contrast and backscattering**

```python
import os
from pathlib import Path
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

data_dir = Path("../data/train/hr")
sample_dir = Path("../sample_images")

# Look for available images
images = list(data_dir.glob("*.png")) + list(data_dir.glob("*.jpg"))
if not images:
    images = list(sample_dir.glob("*.png"))

print(f"Found {len(images)} images for exploration.")
```

### Visualizing RGB Channel Distribution in Underwater Scenes
Underwater images exhibit a pronounced skew where the red channel energy is severely decayed compared to blue and green.

```python
if images:
    sample_path = images[0]
    img = Image.open(sample_path).convert("RGB")
    arr = np.array(img)

    fig, axs = plt.subplots(1, 4, figsize=(16, 4))
    axs[0].imshow(arr)
    axs[0].set_title("Original Underwater Image")
    axs[0].axis("off")

    colors = [('Red Channel', 'Reds'), ('Green Channel', 'Greens'), ('Blue Channel', 'Blues')]
    for i, (name, cmap) in enumerate(colors):
        axs[i+1].imshow(arr[:, :, i], cmap=cmap)
        axs[i+1].set_title(name)
        axs[i+1].axis("off")
    plt.tight_layout()
    plt.show()
```

### Channel Means & Histograms
Quantify mean pixel values per channel across the scene.

```python
if images:
    r_mean = arr[:, :, 0].mean()
    g_mean = arr[:, :, 1].mean()
    b_mean = arr[:, :, 2].mean()

    print(f"Channel Mean Intensity:")
    print(f"  Red:   {r_mean:.2f}")
    print(f"  Green: {g_mean:.2f}")
    print(f"  Blue:  {b_mean:.2f}")

    plt.figure(figsize=(8, 3))
    for i, col in enumerate(['red', 'green', 'blue']):
        plt.hist(arr[:, :, i].ravel(), bins=64, color=col, alpha=0.5, label=f'{col.capitalize()}')
    plt.title("Color Histogram")
    plt.xlabel("Pixel Value")
    plt.ylabel("Count")
    plt.legend()
    plt.show()
```
